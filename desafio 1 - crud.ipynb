```python
pessoa = dict()
cadastros = list()
```

```python
#cadastro
def cadastro_usuario():
  while True:
    pessoa.clear()
    print("-"*30)
    print("MENU CADASTRO: ")
    pessoa['nome'] =str(input("Nome: ")).lower()
    pessoa['telefone'] = int(input("Telefone: "))
    pessoa['email'] = str(input("Email: "))
    print("-"*30)


    cadastros.append(pessoa.copy())

    resp = str(input("Continuar cadastro: S/N")).upper()[0]

    if resp == "N":
        break
    elif resp != "S":
        print("Responda apenas S ou N.")


```

```python
def listar_usuarios():
  print("TABELA DE CADASTROS: ")
  for c in cadastros:
      print("-"*30)
      print(c)
      print("-"*30)


```

```python
def atualizar_usuario():
    procurar = input("Digite nome do usuário: ").lower()
    for use in cadastros:
        if use["nome"] == procurar:
            print("""O que deseja atualizar?
            Nome - (1)
            Telefone - (2)
            Email - (3)
            """)
            opcao = input("Escolha uma opção: ")

            match opcao:
                case "1":
                    atualizar = input("Digite o novo nome: ").lower()
                    use["nome"] = atualizar
                case "2":
                    atualizar = input("Digite o novo telefone: ")
                    use["telefone"] = atualizar
                case "3":
                    atualizar = input("Digite o novo email: ")
                    use["email"] = atualizar
                case _:
                    print(" Opção inválida.")

            print("Cadastro atualizado com sucesso:", use)
            return

```

```python
def deletar_usuario():
    procurar = input("Digite nome do usuario: ").lower()
    for use in cadastros:
      if use["nome"] == procurar:
        cadastros.remove(use)
        print("-"*30)
        print(f"Usuario {procurar} deletado")
        print("-"*30)

```

```python
while True:
    selecionar = int(input("""DIGITE O NÚMERO CORRESPONDENTE:
    CADASTRAR-(1)
    LISTAR-(2)
    ATUALIZAR-(3)
    DELETAR-(4)
    SAIR-(5)
    """))

    match selecionar:
      case 1:
        cadastro_usuario()
      case 2:
        listar_usuarios()
      case 3:
        atualizar_usuario()
      case 4:
        deletar_usuario()
      case 5:
        print("Saindo do programa...")
        break

```
